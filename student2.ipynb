```python
# ===================== Tutorial 2 – Exercise =====================
import json
import os
import ipywidgets as widgets
from IPython.display import display, HTML
from google.colab import drive

drive.mount('/content/drive')

# Path of the file in Drive: My Drive > מחשוב ענן > תרגול 2 > Students.json
FILE_PATH = '/content/drive/MyDrive/מחשוב ענן/תרגול 2/Students.json'

# True = rewrite the file with the original students (erases all saved programs).
# Keep False - the file is created automatically if it does not exist.
WRITE_FILE = False

students_data = {
    "students": [
        {
            "first_name": "Tali",
            "last_name": "Hamoud",
            "braude_email": "Talihamoud@e.braude.ac.il",
            "courses": ["WEB"],
            "interesting_link": "",
            "favorite_programs": []
        },
        {
            "first_name": "Hadi",
            "last_name": "Shofania",
            "braude_email": "HadiShofania@e.braude.ac.il",
            "courses": ["WEB"],
            "interesting_link": "",
            "favorite_programs": []
        },
        {
            "first_name": "Yaren",
            "last_name": "Kablan",
            "braude_email": "YarenKablan@e.braude.ac.il",
            "courses": ["WEB"],
            "interesting_link": "",
            "favorite_programs": []
        },
        {
            "first_name": "Lina",
            "last_name": "Shannan",
            "braude_email": "LinaShnnan@e.braude.ac.il",
            "courses": ["WEB"],
            "interesting_link": "",
            "favorite_programs": []
        }
    ]
}

if WRITE_FILE or not os.path.exists(FILE_PATH):
    with open(FILE_PATH, "w", encoding="utf-8") as fid:
        json.dump(students_data, fid, indent=4, ensure_ascii=False)
    print("Students.json was written successfully")


# ---------- Read / save the JSON file ----------
def load_students():
    with open(FILE_PATH, "r", encoding="utf-8") as fid:
        data = json.load(fid)
    students = data["students"]
    # Convert the old format ("favorite_program": "text") to the new list format
    for s in students:
        if "favorite_programs" not in s:
            old = s.pop("favorite_program", "")
            s["favorite_programs"] = [old] if old else []
        else:
            s.pop("favorite_program", None)
    return students


def save_students(students):
    with open(FILE_PATH, "w", encoding="utf-8") as fid:
        json.dump({"students": students}, fid, indent=4, ensure_ascii=False)


def full_name(student):
    return student["first_name"] + " " + student["last_name"]


students = load_students()
save_students(students)   # save the file in the new format

# ---------- Style ----------
display(HTML("""
<style>
.form-card {
    background: linear-gradient(135deg, #e0f2fe 0%, #f5f3ff 100%);
    border: 2px solid #6366f1;
    border-radius: 14px;
    padding: 18px 22px;
}
.form-title {
    font-family: 'Segoe UI', Arial, sans-serif;
    font-size: 22px;
    font-weight: 700;
    color: #3730a3;
    margin-bottom: 4px;
}
.form-sub {
    font-family: 'Segoe UI', Arial, sans-serif;
    font-size: 13px;
    color: #475569;
    margin-bottom: 10px;
}
.section-title {
    font-family: 'Segoe UI', Arial, sans-serif;
    font-size: 16px;
    font-weight: 700;
    color: #4338ca;
    margin: 4px 0 6px 0;
}
.form-card .widget-label { font-weight: 600; color: #1e293b; }
.fav-list {
    margin: 0 0 0 155px;
    padding: 8px 14px 8px 30px;
    width: 300px;
    background: white;
    border: 1px solid #c7d2fe;
    border-radius: 8px;
    font-family: 'Segoe UI', Arial, sans-serif;
    font-size: 14px;
    color: #1e293b;
}
.fav-list li { padding: 2px 0; }
.fav-label {
    margin-left: 0;
    width: 150px;
    display: inline-block;
    text-align: right;
    font-weight: 600;
    color: #1e293b;
    font-family: 'Segoe UI', Arial, sans-serif;
    font-size: 14px;
}
.fav-table {
    border-collapse: collapse;
    font-family: 'Segoe UI', Arial, sans-serif;
    font-size: 14px;
    min-width: 480px;
}
.fav-table th {
    background: #6366f1;
    color: white;
    padding: 8px 14px;
    text-align: left;
}
.fav-table td {
    background: white;
    color: #1e293b;
    padding: 8px 14px;
    border-bottom: 1px solid #e2e8f0;
    vertical-align: top;
}
.fav-table tr.selected td {
    background: #fef9c3;
    font-weight: 700;
}
.fav-empty { color: #94a3b8; font-style: italic; }
</style>
"""))

label_layout = {'description_width': '150px'}
box_layout = widgets.Layout(width='480px')

# ---------- Widgets ----------
title = widgets.HTML(
    "<div class='form-title'>🎓 Braude Students – Cloud Computing</div>"
    "<div class='form-sub'>Choose a student to see their details, then add or delete favorite programs</div>"
)

student_dropdown = widgets.Dropdown(
    options=[("-- Choose a student --", None)] +
            [(full_name(s), i) for i, s in enumerate(students)],
    value=None,
    description="Student:",
    style=label_layout,
    layout=box_layout
)

# --- Student details section ---
details_title = widgets.HTML("<div class='section-title'>👤 Student Details</div>")

first_name_box = widgets.Text(description="First Name:", disabled=True,
                              style=label_layout, layout=box_layout)
last_name_box = widgets.Text(description="Last Name:", disabled=True,
                             style=label_layout, layout=box_layout)
email_box = widgets.Text(description="Braude Email:", disabled=True,
                         style=label_layout, layout=box_layout)
courses_box = widgets.Textarea(description="Courses:", disabled=True,
                               style=label_layout,
                               layout=widgets.Layout(width='480px', height='70px'))
link_box = widgets.Text(description="Interesting Link:", disabled=True,
                        style=label_layout, layout=box_layout)
link_html = widgets.HTML()
programs_html = widgets.HTML()   # list of the student's favorite programs

# --- Add section ---
add_title = widgets.HTML("<div class='section-title'>➕ Add Favorite Program</div>")

new_program_box = widgets.Text(description="New Program:",
                               placeholder="Type a favorite program and press Update",
                               style=label_layout, layout=box_layout)

update_button = widgets.Button(description="Update",
                               button_style="success",
                               icon="save",
                               layout=widgets.Layout(width='150px', margin='10px 0 0 155px'))

# --- Delete section ---
delete_title = widgets.HTML("<div class='section-title'>🗑 Delete Favorite Program</div>")

delete_dropdown = widgets.Dropdown(options=[("-- No programs --", None)],
                                   value=None,
                                   description="Choose Program:",
                                   style=label_layout,
                                   layout=box_layout)

delete_button = widgets.Button(description="Delete",
                               button_style="danger",
                               icon="trash",
                               layout=widgets.Layout(width='150px', margin='10px 0 0 155px'))

status_label = widgets.HTML()

# --- All favorite programs section ---
table_title = widgets.HTML("<div class='section-title'>📺 Favorite Programs of All Students</div>")
table_html = widgets.HTML()


# ---------- Logic ----------
def refresh_table():
    """Build the table of all students and their favorite programs."""
    selected = student_dropdown.value
    rows = ""
    for i, s in enumerate(students):
        programs = s["favorite_programs"]
        if programs:
            program_cell = "<br>".join(f"• {p}" for p in programs)
        else:
            program_cell = "<span class='fav-empty'>— none —</span>"
        row_class = " class='selected'" if i == selected else ""
        rows += f"<tr{row_class}><td>{full_name(s)}</td><td>{program_cell}</td></tr>"
    table_html.value = (f"<table class='fav-table'>"
                        f"<tr><th>Student</th><th>Favorite Programs</th></tr>"
                        f"{rows}</table>")


def refresh_programs():
    """Show the chosen student's programs and fill the delete dropdown."""
    index = student_dropdown.value
    if index is None:
        programs_html.value = ""
        delete_dropdown.options = [("-- No programs --", None)]
        delete_dropdown.value = None
        return

    programs = students[index]["favorite_programs"]
    if programs:
        items = "".join(f"<li>{p}</li>" for p in programs)
        programs_html.value = (f"<div class='fav-label'>⭐ Favorite Programs:</div>"
                               f"<ol class='fav-list'>{items}</ol>")
        delete_dropdown.options = ([("-- Choose a program to delete --", None)] +
                                   [(p, i) for i, p in enumerate(programs)])
    else:
        programs_html.value = ("<div class='fav-label'>⭐ Favorite Programs:</div>"
                               "<div class='fav-list fav-empty'>No favorite programs yet</div>")
        delete_dropdown.options = [("-- No programs --", None)]
    delete_dropdown.value = None


def clear_fields():
    for box in (first_name_box, last_name_box, email_box, courses_box,
                link_box, new_program_box):
        box.value = ""
    link_html.value = ""


def on_student_change(change):
    status_label.value = ""
    index = change["new"]
    if index is None:
        clear_fields()
        refresh_programs()
        refresh_table()
        return
    student = students[index]
    first_name_box.value = student["first_name"]
    last_name_box.value = student["last_name"]
    email_box.value = student["braude_email"]
    courses_box.value = "\n".join(student["courses"])
    link_box.value = student.get("interesting_link", "")
    if link_box.value:
        link_html.value = (f"<div style='margin-left:155px'>"
                           f"<a href='{link_box.value}' target='_blank'>🔗 Open link</a></div>")
    else:
        link_html.value = ""
    new_program_box.value = ""
    refresh_programs()
    refresh_table()


def on_update_click(button):
    index = student_dropdown.value
    if index is None:
        status_label.value = "<b style='color:#dc2626'>⚠ Please choose a student first.</b>"
        return
    new_program = new_program_box.value.strip()
    if new_program == "":
        status_label.value = "<b style='color:#dc2626'>⚠ Please type a favorite program.</b>"
        return

    programs = students[index]["favorite_programs"]
    if new_program.lower() in [p.lower() for p in programs]:
        status_label.value = (f"<b style='color:#d97706'>ℹ '{new_program}' is already in "
                              f"{full_name(students[index])}'s list.</b>")
        return

    # Add the program to the chosen student's list and save to the JSON file
    programs.append(new_program)
    save_students(students)

    new_program_box.value = ""
    refresh_programs()
    refresh_table()

    status_label.value = (f"<b style='color:#16a34a'>✔ Added '{new_program}' to "
                          f"{full_name(students[index])}'s favorite programs "
                          f"(saved to Students.json)</b>")


def on_delete_click(button):
    index = student_dropdown.value
    if index is None:
        status_label.value = "<b style='color:#dc2626'>⚠ Please choose a student first.</b>"
        return
    programs = students[index]["favorite_programs"]
    if not programs:
        status_label.value = (f"<b style='color:#d97706'>ℹ {full_name(students[index])} "
                              f"has no favorite programs to delete.</b>")
        return
    program_index = delete_dropdown.value
    if program_index is None:
        status_label.value = "<b style='color:#dc2626'>⚠ Please choose a program to delete.</b>"
        return

    # Remove the chosen program from the list and save to the JSON file
    removed = programs.pop(program_index)
    save_students(students)

    refresh_programs()
    refresh_table()

    status_label.value = (f"<b style='color:#dc2626'>🗑 Deleted '{removed}' from "
                          f"{full_name(students[index])}'s favorite programs "
                          f"(saved to Students.json)</b>")


student_dropdown.observe(on_student_change, names="value")
update_button.on_click(on_update_click)
delete_button.on_click(on_delete_click)

refresh_table()

# ---------- Layout ----------
divider = "<hr style='border:1px solid #c7d2fe'>"

form = widgets.VBox([
    title,
    student_dropdown,
    widgets.HTML(divider),
    details_title,
    first_name_box,
    last_name_box,
    email_box,
    courses_box,
    link_box,
    link_html,
    programs_html,
    widgets.HTML(divider),
    add_title,
    new_program_box,
    update_button,
    widgets.HTML(divider),
    delete_title,
    delete_dropdown,
    delete_button,
    widgets.HTML(divider),
    status_label,
    table_title,
    table_html
])
form.add_class("form-card")

display(form)
```
